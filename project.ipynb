# Exploring Hacker News Posts

With this [dataset](https://www.kaggle.com/hacker-news/hacker-news-posts) on Hacker News, I will look into the posts of Show HN and Ask HN by exploring this given file

```python
from csv import reader

openedFile = open("hacker_news.csv", encoding='utf8')
readFile = reader(openedFile)
hn = list(readFile)
openedFile.close()

for row in hn[:5]:
    print(row)
```

```python
headers = hn[0]
hn = hn[1:]

print(headers)
print(hn[:5])
```

```python
ask_posts = []
show_posts = []
other_posts = []

for row in hn:
    title = row[1]
    if (title.lower()).startswith("ask hn"):
        ask_posts.append(row)
    elif (title.lower()).startswith("show hn"):
        show_posts.append(row)
    else:
        other_posts.append(row)

print(len(ask_posts))
print(len(show_posts))
print(len(other_posts))
```

```python
total_ask_comments = 0

for row in ask_posts:
    commentNum = int(row[4])
    total_ask_comments += commentNum

avg_ask_comments = total_ask_comments / len(ask_posts)
print(avg_ask_comments)

total_show_comments = 0

for row in show_posts:
    commentNum = int(row[4])
    total_show_comments += commentNum

avg_show_comments = total_show_comments / len(show_posts)
print(avg_show_comments)
```

## Do show posts or ask posts receive more comments on average?

Ask posts have more comments in general. This may be because people are answering these questions instead of just admiring it.

```python
import datetime as dt

result_list = []

for row in ask_posts:
    created_at = row[6]
    commentNum = int(row[4])
    result_list.append([created_at, commentNum])

counts_by_hour = {}
comments_by_hour = {}

for row in result_list:
    #date and time in '9/26/2016 3:26' this format
    dateAndTime = dt.datetime.strptime(row[0], "%m/%d/%Y %H:%M")
    timeHour = dateAndTime.strftime("%H")
    
    if timeHour in counts_by_hour:
        counts_by_hour[timeHour] += 1
        comments_by_hour[timeHour] += row[1]
    else:
        counts_by_hour[timeHour] = 1
        comments_by_hour[timeHour] = row[1]

print(counts_by_hour)
print(comments_by_hour)
```

```python
avg_by_hour = []

for key in counts_by_hour:
    timeHour = key
    avgCommentsPerPost = comments_by_hour[key] / counts_by_hour[key]
    avg_by_hour.append([timeHour, avgCommentsPerPost])

avg_by_hour.sort()
print(avg_by_hour)
```

```python
swap_avg_by_hour = []

for row in avg_by_hour:
    swap_avg_by_hour.append([row[1], row[0]])

print(swap_avg_by_hour)

sorted_swap = sorted(swap_avg_by_hour, reverse=True)

print("Top 5 Hours for Ask Posts Comments")

template = "{}:00 {:.2f} average comments per post"
for row in sorted_swap[:5]:
    timeHour = dt.datetime.strptime(row[1], "%H")
    output = template.format(timeHour.hour, row[0])
    print(output)
```

The highest chances of getting you post noticed is around 12pm-3pm as this time period has the highest average comments per post
